## ONLINE SHOPPING CART SYSTEM

```python
def shopping_cart_system():

    product_list = ["Laptop","Smartphone","Headphone","Smartwatch","Backpack"]
    
    product_tuple = (
        ("Laptop",45000),
        ("Smartphone",35000),
        ("Headphone",5000),
        ("Smartwatch",3000),
        ("Backpack",1500)
    )
    price_catalog = dict(product_tuple)

    categories_set = {"electronics","accessories","wearables","bags"}

    cart = {}

    print("=" * 50)
    print("WELCOME TO ONLINE SHOPPING CART SYSTEM")
    print("=" * 50)

    print("\nAvaliable product = ")
    for index,product in enumerate(product_list,start = 1):
        price = price_catalog[product]
        print(f"{index}.{product} - ₹{price}")

    print(f"\nProduct Categories: {', '.join(categories_set)}")
    print("-" * 50)

    while True:
        print("\nOptions: ")
        print("1. Add product to cart")
        print("2. View cart and checkout(exit)")

        choice = input("enter your choice(1 or 2): ").strip()

        if choice == '1':
            product_name = input("Enter the product name you want to buy: ").strip().title()

            if product_name not in product_list:
                print("Error:Selected product is unavaliable.Please choose from the list.")
                continue
            try:
                quantity = int(input(f"enter quantity for {product_name}: "))
            except ValueError:
                print("Error:please enter a valid numeric quantity.")
                continue

            if quantity <= 0:
                print("Error:quantity cannot be zero or negative.")
                continue

            if product_name in cart:
                cart[product_name] += quantity
            else:
                cart[product_name] = quantity
            print(f"successfully added {quantity} x {product_name} into your cart.")    

        elif choice == '2':
            break
        else:
            print("invalid choice.Please enter 1 or 2.")

    print("\n"+"=" * 50)
    print("FINAL BILL SUMMARY")
    print("=" * 50)  

    if not cart:
        print("Your cart is empty.Thank you for visiting!")
        return

    subtotal = 0
    print(f"{'product':<15} {'quantity':<8} {'price':<10} {'total'}") 
    print("-" * 50)

    for product, quantity in cart.items():
        unit_price = price_catalog[product]
        item_total = unit_price * quantity
        subtotal += item_total
        print(f"{product:<15} {quantity:<8} ₹{unit_price:<9} ₹{item_total}")

    print("-" * 50)
    print(f"subtotal:₹{subtotal}")

    discount = 0
    if subtotal >= 5000:
        discount = subtotal * 0.10
        print(f"Discount (10%): -₹{discount}")

    else:
        print("Discount: ₹0 (Shop for ₹5,000 or more to get 10% off)")

    final_amount = subtotal - discount
    print("=" * 50)
    print(f"Final Amount to Pay: ₹{final_amount}")
    print("=" * 50)
    print("Thank you for shopping with us!")

# Run the program
if __name__ == "__main__":
    shopping_cart_system()
```
